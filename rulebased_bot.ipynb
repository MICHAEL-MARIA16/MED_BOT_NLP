```python
import random
import re

class MedBot:
    neg_responses = ("no", "nope", "nah", "not a chance", "sorry")
    exitcomm = ("quit", "pause", "exit", "goodbye", "bye", "later")
    
    randomques = (
        "I have a headache. What natural remedy can I try?",
        "What should I do for a sore throat without using medicine?",
        "Can I treat an upset stomach with herbal remedies?",
        "What home remedy helps relieve back pain?",
        "Are there any natural ways to boost my immunity?",
        "How can I reduce stress using natural methods?",
        "What are good home remedies for acne?",
        "How do I naturally treat dry skin?",
        "What should I drink to help with a cold?",
        "Which natural foods help with digestion?"
    )

    def __init__(self):
        self.intents = {
            'describe_remedy_intent': r'.*\s*(remedy|cure|treatment).*',
            'ask_headache_remedy': r'.*\s*headache.*',
            'ask_sore_throat_remedy': r'.*\s*sore throat.*',
            'ask_stomach_remedy': r'.*\s*stomach.*',
            'ask_back_pain_remedy': r'.*\s*back pain.*'
        }

    def greet(self):
        self.name = input("Hey, what's your name?\n")
        will_help = input(f"Hi {self.name}, I am MedBot created to help you with natural remedies. Would you like to try it?\n")
        if will_help.lower() in self.neg_responses:
            print("Okay, have a great day! Catch you later!")
            return
        self.chat()

    def make_exit(self, reply):
        for command in self.exitcomm:
            if reply == command:
                print("Okay, have a great day! Take care!")
                return True
        return False

    def chat(self):
        reply = input(random.choice(self.randomques)).lower()
        while not self.make_exit(reply):
            reply = input(self.match_reply(reply)).lower()

    def match_reply(self, reply):
        for key, pattern in self.intents.items():
            found_match = re.match(pattern, reply)
            if found_match and key == 'ask_headache_remedy':
                return self.suggest_headache_remedy()
            elif found_match and key == 'ask_sore_throat_remedy':
                return self.suggest_sore_throat_remedy()
            elif found_match and key == 'ask_stomach_remedy':
                return self.suggest_stomach_remedy()
            elif found_match and key == 'ask_back_pain_remedy':
                return self.suggest_back_pain_remedy()
        return self.no_match_intent()

    def suggest_headache_remedy(self):
        responses = [
            "Try drinking water or having ginger tea to reduce your headache.",
            "Peppermint oil or a cold compress on your forehead can help relieve headaches."
        ]
        return random.choice(responses)

    def suggest_sore_throat_remedy(self):
        responses = [
            "Gargling with warm salt water can help soothe a sore throat.",
            "Drinking warm honey lemon tea is great for easing throat pain."
        ]
        return random.choice(responses)

    def suggest_stomach_remedy(self):
        responses = [
            "Ginger tea or chewing on fennel seeds can help soothe an upset stomach.",
            "A banana or plain yogurt might ease your stomach discomfort."
        ]
        return random.choice(responses)

    def suggest_back_pain_remedy(self):
        responses = [
            "Try a warm compress or a gentle back massage with essential oils for back pain relief.",
            "Stretching and improving posture can also help alleviate back pain."
        ]
        return random.choice(responses)

    def no_match_intent(self):
        responses = [
            "I'm not sure I understand. Can you tell me more?",
            "Could you clarify your question, please?"
        ]
        return random.choice(responses)

# Create the MedBot instance and start the conversation
medbot = MedBot()
medbot.greet()
```
